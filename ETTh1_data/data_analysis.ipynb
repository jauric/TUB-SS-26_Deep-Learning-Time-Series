```python
import matplotlib.dates as mdates
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
ETTh1_link = "ETTh1.csv"

data_ETTh1 = pd.read_csv(ETTh1_link)
data_ETTh1.head()
```

```python
figure, axes = plt.subplots(7, 1, figsize=(20, 20), sharex=True)

data_ETTh1.plot(x="date", y="HUFL", ax=axes[0], title="HUFL")
data_ETTh1.plot(x="date", y="HULL", ax=axes[1], title="HULL")
data_ETTh1.plot(x="date", y="MUFL", ax=axes[2], title="MUFL")
data_ETTh1.plot(x="date", y="MULL", ax=axes[3], title="MULL")
data_ETTh1.plot(x="date", y="LUFL", ax=axes[4], title="LUFL")
data_ETTh1.plot(x="date", y="LULL", ax=axes[5], title="LULL")
data_ETTh1.plot(x="date", y="OT", ax=axes[6], title="Oil Temp")

plt.tight_layout()
plt.show()
```
