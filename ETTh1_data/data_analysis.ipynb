```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# Import from Google Drive as DataFrame

ETTh1_url = "https://drive.google.com/uc?export=download&id=1mtstKglYRWTngqyJ1-Yfq4mwdKHucRAa"       

# NOTES on this data_url
# - Data stored on google drive can be imported with a shared link: right click on file -> share -> share. In the pop-up under "General access", change access from "Restricted" to "Anyone with the link" and copy the link
# - This link has the form: https://drive.google.com/file/d/FILE_ID/view?usp=sharing --> copy the FILE_ID
# - PROBLEM: directly loading the csv from the shared link above can lead to errors because Google Drive doesn't provide the csv directly but often just an html file, breaking proper import with pandas
# - SOLUTION: use the file download link instead, it has the form: https://drive.google.com/uc?export=download&id=FILE_ID --> use this link template and replace FILE_ID with the characters copied from the shared link above
# - You can now properly import the csv file with pandas as a DataFrame

data_ETTh1 = pd.read_csv(ETTh1_url, sep = ",")
data_ETTh1.head()
```

```python
figure, axes = plt.subplots(1, 2, figsize=(10, 4))

data_ETTh1.plot(x="date", y="HUFL", ax=axes[0], title="HUFL")
data_ETTh1.plot(x="date", y="HUFL", ax=axes[1], title="HULL")

plt.tight_layout()
plt.show()



```
