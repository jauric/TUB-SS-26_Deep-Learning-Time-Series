```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
ETTh1_link = "ETTh1_data/ETTh1.csv"

data_ETTh1 = pd.read_csv(ETTh1_link, sep = ",")
data_ETTh1.head()
```

```python
figure, axes = plt.subplots(1, 2, figsize=(10, 4))

data_ETTh1.plot(x="date", y="HUFL", ax=axes[0], title="HUFL")
data_ETTh1.plot(x="date", y="HUFL", ax=axes[1], title="HULL")

plt.tight_layout()
plt.show()



```
