# **Baseline for ETTh1 | Auto_ARIMA**

- ARIMA = **A**uto**R**egress**I**ve **M**oving-**A**verage
- https://en.wikipedia.org/wiki/Autoregressive_moving-average_model

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

#!pip install pmdarima              # run only the first time to install the package
from pmdarima import auto_arima
from google.colab import files
```

```python
# Import full data set

# ETTh1_link = "ETTh1.csv"
ETTh1_link =  "https://raw.githubusercontent.com/jauric/TUB-SS-26_Deep-Learning-Time-Series/refs/heads/main/ETTh1_data/ETTh1.csv"

data_ETTh1 = pd.read_csv(ETTh1_link, sep = ",")
data_ETTh1.head()
```

```python
# Auto ARIMA model for ETTh1 data set

y = data_ETTh1.set_index("date")["OT"]

model = auto_arima(
    y,
    seasonal=False,
    stepwise=True,
    suppress_warnings=True
)

forecast = model.predict(n_periods=96)
print(forecast)
```
